## 1320. Minimum Distance to Type a Word Using Two Fingers (Hard) [題目](https://leetcode.com/problems/minimum-distance-to-type-a-word-using-two-fingers/description/?envType=daily-question&envId=2026-04-12)

You have a keyboard layout as shown above in the X-Y plane, where each English uppercase letter is located at some coordinate.

For example, the letter 'A' is located at coordinate (0, 0), the letter 'B' is located at coordinate (0, 1),

the letter 'P' is located at coordinate (2, 3) and the letter 'Z' is located at coordinate (4, 1).

Given the string word, return the minimum total distance to type such string using only two fingers.

The distance between coordinates (x1, y1) and (x2, y2) is |x1 - x2| + |y1 - y2|.

Note that the initial positions of your two fingers are considered free so do not count towards your total distance, also your two fingers do not have to start at the first letter or the first two letters.

Example 1:

Input: word = "CAKE"

Output: 3

Explanation: Using two fingers, one optimal way to type "CAKE" is: 
* Finger 1 on letter 'C' -> cost = 0 
* Finger 1 on letter 'A' -> cost = Distance from letter 'C' to letter 'A' = 2 
* Finger 2 on letter 'K' -> cost = 0 
* Finger 2 on letter 'E' -> cost = Distance from letter 'K' to letter 'E' = 1 
* Total distance = 3

```python
class Solution:
    def minimumDistance(self, word: str) -> int:
        # A~Z分別為0~25
        # 定義初始字符為26        
        n = len(word)
        empty_char = 26
        nums = [ord(c) - ord('A') for c in word]
        
        # 建立一個3*27*27的list
        memo = [[[-1] * 27 for _ in range(27)] for _ in range(n)]

        def get_distance(p1, p2):
            if p1 == empty_char: return 0
            row1, col1 = divmod(p1, 6)
            row2, col2 = divmod(p2, 6)
            return abs(row1 - row2) + abs(col1 - col2)
        
        def dp(i, l, r):
            if i == n:
                return 0
            if memo[i][l][r] != -1:
                return memo[i][l][r]
            
            target = nums[i]

            # 計算左手距離
            cost_l = get_distance(l, target) + dp(i+1, target, r)

            # 計算右手距離
            cost_r = get_distance(r, target) + dp(i+1, l, target)

            memo[i][l][r] = min(cost_l, cost_r)
            
            return memo[i][l][r]

        return dp(0, empty_char, empty_char)
```
