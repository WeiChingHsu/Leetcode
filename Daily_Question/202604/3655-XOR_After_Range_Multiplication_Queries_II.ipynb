## 3655. XOR After Range Multiplication Queries II (Hard) [題目](https://leetcode.com/problems/xor-after-range-multiplication-queries-ii/description/?envType=daily-question&envId=2026-04-09)

You are given an integer array nums of length n and a 2D integer array queries of size q, where queries[i] = [li, ri, ki, vi].

For each query, you must apply the following operations in order:

Set idx = li.

While idx <= ri:

Update: nums[idx] = (nums[idx] * vi) % (109 + 7)

Set idx += ki.

Return the bitwise XOR of all elements in nums after processing all queries.

Example 1:

Input: nums = [1,1,1], queries = [[0,2,1,4]]

Output: 4

Explanation:

* A single query [0, 2, 1, 4] multiplies every element from index 0 through index 2 by 4.

* The array changes from [1, 1, 1] to [4, 4, 4].

* The XOR of all elements is 4 ^ 4 ^ 4 = 4.

Constraints:
* 1 <= n == nums.length <= 105
* 1 <= nums[i] <= 109
* 1 <= q == queries.length <= 105​​​​​​​
* queries[i] = [li, ri, ki, vi]
* 0 <= li <= ri < n
* 1 <= ki <= n
* 1 <= vi <= 105

必備知識：乘法差分、費馬小定理

```python
class Solution:
    def xorAfterQueries(self, nums: List[int], queries: List[List[int]]) -> int:
        MOD = 1000000007
        n = len(nums)
        threshold = int(n ** 0.5)

        # 用來存較小步伐的查詢，k從0~B
        small_k_map = [[] for _ in range(threshold + 1)]

        # 區分大步伐和小步伐
        for l, r, k, v in queries:
            if v == 1: continue
            if k > threshold:
                # 對於大步伐直接更新
                for i in range(l, r + 1, k):
                    nums[i] = (nums[i] * v) % MOD
            else:
                small_k_map[k].append((l,r,v))
        
        for k in range(1, threshold + 1):
            if not small_k_map[k]: continue
            
            # 乘法差分初質為1
            # 用空間換取程式碼簡潔，長度設為n+1也行，但程式碼會變複雜
            diff = [1] * (n + k + 1)
            for l, r, v in small_k_map[k]:
                v_invert = pow(v, MOD-2, MOD)
                diff[l] = (diff[l] * v) % MOD # 對l乘上v
                # (r-l) // k 為[l,r]區間的跳躍次數，((r-l) // k) + 1 次跳躍即可跳出區間
                next_pos = l + ((r-l) // k + 1) * k
                if next_pos < n + k:
                    diff[next_pos] = (diff[next_pos] * v_invert) % MOD # 對於next_pos乘上v的乘法反

            for i in range(k, n):
                diff[i] = (diff[i] * diff[i-k]) % MOD
            
            for i in range(n):
                if diff[i] != 1:
                    nums[i] = (nums[i] * diff[i]) % MOD
            
        result = 0
        for x in nums:
            result ^= x
        return result
```
