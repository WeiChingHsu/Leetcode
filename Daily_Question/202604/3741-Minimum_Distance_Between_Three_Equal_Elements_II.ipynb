## 3741. Minimum Distance Between Three Equal Elements II (Medium) [題目](https://leetcode.com/problems/minimum-distance-between-three-equal-elements-i/description/?envType=daily-question&envId=2026-04-10)

You are given an integer array nums.

A tuple (i, j, k) of 3 distinct indices is good if nums[i] == nums[j] == nums[k].

The distance of a good tuple is abs(i - j) + abs(j - k) + abs(k - i), where abs(x) denotes the absolute value of x.

Return an integer denoting the minimum possible distance of a good tuple. If no good tuples exist, return -1.

Example 1:

Input: nums = [1,2,1,1,3]

Output: 6

Explanation:

The minimum distance is achieved by the good tuple (0, 2, 3).

(0, 2, 3) is a good tuple because nums[0] == nums[2] == nums[3] == 1. Its distance is abs(0 - 2) + abs(2 - 3) + abs(3 - 0) = 2 + 1 + 3 = 6.

Constraints:
* 1 <= n == nums.length <= 105
* 1 <= nums[i] <= n

```python
from typing import List
from collections import defaultdict
class Solution:
    def minimumDistance(self, nums: List[int]) -> int:
        minDistance = float("inf")
        pos_map = defaultdict(list) # 存取不存在的key值時不會報錯
        for idx, val in enumerate(nums): # O(n)
            pos_map[val].append(idx)
        
        for indices in pos_map.values(): # 雖然是雙層迴圈，但每個index只會被遍歷一次，故為O(n)
            if len(indices) < 3: continue
            for m in range(len(indices) - 2):
                # 因為這邊的i << j << k
                # 距離公式可簡化成 (j-i) + (k-j) + (k-i) => 2(k-i)
                distance = 2 * (indices[m + 2] - indices[m])
                if distance < minDistance: minDistance = distance

        return int(minDistance) if minDistance != float("inf") else -1
```
