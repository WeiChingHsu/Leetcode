## 3289. The Two Sneaky Numbers of Digitville (Easy)

In the town of Digitville, there was a list of numbers called nums containing integers from 0 to n - 1. Each number was supposed to appear exactly once in the list, however, two mischievous numbers sneaked in an additional time, making the list longer than usual.

As the town detective, your task is to find these two sneaky numbers. Return an array of size two containing the two numbers (in any order), so peace can return to Digitville.

Example 1:

* Input: nums = [0,1,1,0]

* Output: [0,1]

Explanation:

* The numbers 0 and 1 each appear twice in the array.

Example 2:

* Input: nums = [0,3,2,1,3,2]

* Output: [2,3]

Explanation:

* The numbers 2 and 3 each appear twice in the array.

### 1. 直覺Dict法

Time: O(N)<br>
Space: O(N)

```python
class Solution:
    def getSneakyNumbers(self, nums: List[int]) -> List[int]:
        mapp = {}
        result = []
        for item in nums:
            if not item in mapp:
                mapp[item] = 1
            else:
                result.append(item)
        return result
```

### 2. XOR法 (神級解法)

Time: O(N)<br>
Space: O(1)

```python
class Solution:
    def getSneakyNumbers(self, nums: List[int]) -> List[int]:
        n = len(nums) - 2
        xor_sum = 0

        # 因為兩個一樣的數做XOR會得到0
        # 故xor_sum即為所求[a, b]做了XOR的值
        for x in nums:
            xor_sum ^= x
        for i in range(n):                                      
            xor_sum ^= i

        # -xor_sum = ~xor_sum + 1
        # 對xor_sum和-xor_sum做and operation可以得到xor-sum最右邊的1的位子(只有一個位元是1，其餘皆為0)
        mask = xor_sum & -xor_sum

        a, b = 0, 0
        for x in nums:
            if x & mask:
                a ^= x
            else:
                b ^= x

        for i in range(n):
            if i & mask:
                a ^= i
            else:
                b ^= i
        # 由於通過mask的正常數字必為2個，根據XOR性質，A XOR A = 0，故最後剩下烙單的a,b
        return [a, b]
```
